# Introduccón

Se comienza definiendo una función que nos permita generar una señal senoidal. Para ello, hay que tener en cuanta que la forma general para una señal senoidal es: $$ x(t) = A.sin(2\pi ft+\phi) + DC$$

donde:

- $A$ $\rightarrow$ amplitud máxima

- $f$ $\rightarrow$ frecuencia (Hz)

- $t$ $\rightarrow$ tiempo

- $\phi$ $\rightarrow$ fase inicial

- $DC$ $\rightarrow$ valor medio (offset)


```python
# Bloque para crear la función pedida
import numpy as np

def mi_funcion_sen(vmax, dc, ff, ph, nn, fs):

    # tiempo de muestreo
    ts = 1/fs
    
    # vector de tiempo
    tt = np.linspace(0, (nn-1)*ts, nn).reshape(nn,1) # El reshape es para que sean vectores de Nx1
    
    # señal senoidal
    xx = vmax * np.sin(2*np.pi*ff*tt + ph) + dc
    
    return tt, xx
```

```python
# Bloque de ejemplo de implementación
import matplotlib.pyplot as plt

N = 1000
fs = 1000

tt, xx = mi_funcion_sen(vmax=1, dc=0, ff=1, ph=0, nn=N, fs=fs)

plt.plot(tt, xx,color='#00FF00')
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.title("Señal senoidal")
plt.grid(True)

plt.show()

# Verifico que que los vectores tt y xx sean de la forma Nx1

print('Vector tt:', tt.shape)
print('Vector xx:', xx.shape)
```
