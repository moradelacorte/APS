# Introduccón

Se comienza definiendo una función que nos permita generar una señal senoidal. Para ello, hay que tener en cuanta que la forma general para una señal senoidal es: $$ x(t) = A.sin(2\pi ft+\phi) + DC$$

donde:

- $A$ $\rightarrow$ amplitud máxima

- $f$ $\rightarrow$ frecuencia (Hz)

- $t$ $\rightarrow$ tiempo

- $\phi$ $\rightarrow$ fase inicial

- $DC$ $\rightarrow$ valor medio (offset)


## Generando una señal senoidal

```python
# Bloque para crear la función pedida
import numpy as np

def mi_funcion_sen(vmax, dc, ff, ph, nn, fs):

    # tiempo de muestreo
    ts = 1/fs
    
    # vector de tiempo
    tt = np.linspace(0, (nn-1)*ts, nn).reshape(nn,1) # El reshape es para que sean vectores de Nx1
    
    # señal senoidal
    xx = vmax * np.sin(2*np.pi*ff*tt + ph) + dc
    
    return tt, xx
```

```python
# Bloque de ejemplo de implementación
import matplotlib.pyplot as plt

N = 1000
fs = 1000

tt, xx = mi_funcion_sen(vmax=1, dc=0, ff=1, ph=0, nn=N, fs=fs)

plt.plot(tt, xx,color='#00FF00')
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.title("Señal senoidal")
plt.grid(True)

plt.show()

# Verifico que que los vectores tt y xx sean de la forma Nx1

print('Vector tt:', tt.shape)
print('Vector xx:', xx.shape)
```

# Otros tipos de señales

## Generando una señal cuadrada

```python
# Bloque para crear la función
import numpy as np
import matplotlib.pyplot as plt

def mi_funcion_square(vmax, dc, ff, ph, nn, fs):

    # tiempo de muestreo
    ts = 1/fs
    
    # vector de tiempo
    tt = np.linspace(0, (nn-1)*ts, nn).reshape(nn,1) # El reshape es para que sean vectores de Nx1

    xx = np.sign(vmax * np.sin(2*np.pi*ff*tt + ph) + dc)
    
    return tt,xx
```

La función **np.sign()** me da como resultado: 
- +1 si (lo del parentesis) > 0 
- -1 si (lo del parentesis) < 0

*Ejemplo:*

```python
a = 1

b = -3

print(np.sign(a+b)) #devuelve -1

print(np.sign(a-b)) #devuelve +1

print(np.sign(a-1)) #devuelve 0
```

```python
# Bloque de ejemplo de implementación
import matplotlib.pyplot as plt

N = 1000
fs = 1000
ff = 1001

tt, xx = mi_funcion_square(vmax=1, dc=0, ff=ff, ph=0, nn=N, fs=fs)

plt.plot(tt, xx,color='#00FF00')
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.title("Señal senoidal")
plt.grid(True)

plt.show()
```

## Generando una señal con ruido

Para esta parte usamos el concepto de *SNR = Signal to Noise Ratio*, que mide qué tan fuerte es la señal en comparación con el ruido presente en ella.

Se define como: $$SNR=P_{señal}/P_{ruido}$$
donde
- $P_{señal}$ $\rightarrow$ Potencia de la señal
- $P_{ruido}$ $\rightarrow$ Potencia del ruido

El la practica suele expresarse en decibeles (dB): $$SNR_{dB}=10\log_{10}(P_{señal}/P_{ruido})$$

Si se conoce el valor para $SNR_{dB}$ , es posible despejar el valor para la potencia del ruido como $$P_{ruido}=P_{señal}/10^{SNR_{dB}/10}$$ 

En este trabajo se modela el ruido como ruido blanco gaussiano, el cual cumple que su potencia es igual a su varianza: $$Var(ruido)=\sigma^2$$

Por lo tanto, si conocemos la potencia del ruido, el desvío estándar del ruido queda dado por: $$\sigma = \sqrt{P_{ruido}}$$

Este valor de $\sigma$ se utiliza para escalar el ruido gaussiano generado y luego sumarlo a la señal senoidal simulada.

```python
vmax = np.sqrt(2)
ff = 1
N = 100
fs = 100
snr = 20 #SNR=20dB: señal 100 veces más potente que el ruido

tt, xx = mi_funcion_sen(vmax=vmax, dc=0, ff=ff, ph=0, nn=N, fs=fs)

ps = np.var(xx) # Uso la variancia para calcular la potencia de la señal
pr = 10**-(snr/10) # Despejando de la ecuacion para SNR_dB

ruido = np.sqrt(pr) * np.random.normal(0,1,(N,1))

yy = xx + ruido

import matplotlib.pyplot as plt

plt.plot(tt,xx,label="señal limpia")
plt.plot(tt,yy,label="señal con ruido")

plt.xlabel("tiempo")
plt.ylabel("Voltaje")
plt.title("Senoidal con ruido")
plt.legend()

plt.show()

print('Potencia de la señal: ', ps)
print('Potencia del ruido: ', pr)
```
